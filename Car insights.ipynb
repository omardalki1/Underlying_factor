# Does engine size corelate with horsepower? 

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r'C:\Users\owner\Downloads\Car_sales.csv',na_values = ['.','x','X'])
df.fillna(0,inplace = True)
print(df.head())
```

```python
#step = len(df) // 10
df[["Horsepower","Engine size"]] = df[["Horsepower","Engine size"]].astype(float)  
plt.scatter(df["Horsepower"],df["Engine size"],label = "Are these two related?",color = 'blue',marker = 'o')
plt.title("the underlying factor")
plt.xlabel("Horse power")
plt.ylabel("Engine size")
#plt.xticks(ticks = df["Engine size"][::step],rotation = 45)
#plt.yticks(ticks = df["Horsepower"][::step],rotation = 45)
plt.show()



```
